```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
car=pd.read_csv("car.csv")
```

```python
car
```

```python
car.isnull().sum()
```

```python
sns.heatmap(car.isnull())
```

## Use Machine learning
## Divide the Data into training and testing


```python
from sklearn.model_selection import train_test_split
```

```python
#As we want the data to be numerical type such that we are changing the colums in in the numerical type
car['Fuel_Type'] = car['Fuel_Type'].map({
    'Petrol': 0,
    'Diesel': 1,
    'CNG': 2
})
```

```python
car['Seller_Type'] = car['Seller_Type'].map({
    'Dealer': 0,
    'Individual': 1
})
```

```python
car['Transmission'] = car['Transmission'].map({
    'Manual': 0,
    'Automatic': 1
})
```

```python
car
```

```python
X = car.drop(["Selling_Price","Car_Name"], axis=1)
y = car['Selling_Price']
X_train, X_test, y_train, y_test = train_test_split(X, y,test_size=0.22,random_state=3)
```

```python
#The independent paramenters and dependent parameters are taken individually
X
```

## Random Forest Regressor

```python
from sklearn.ensemble import RandomForestRegressor

# By using Random Forest Regressor we can get better prediction
rf = RandomForestRegressor(n_estimators=200, random_state=42)
rf.fit(X_train, y_train)

y_pred = rf.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error
# MAE
mae = mean_absolute_error(y_test, y_pred)
print("MAE :", mae)

```

```python
from sklearn.metrics import mean_squared_error
# MSE
mse = mean_squared_error(y_test, y_pred)
print("MSE :", mse)
```

```python
from sklearn.metrics import  r2_score
# R2 Score
r2 = r2_score(y_test, y_pred)
print("R2  :", r2)
```

```python
#plotting the predicted values to original values
plt.scatter(y_test, y_pred)
plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.title("Actual vs Predicted")
plt.show()
```

## predecting the Selling price

```python
import numpy as np
import warnings
from datetime import datetime
showroom_price = float(input("Enter showroom price (in Lakhs): "))
manufacture_year = int(input("Enter manufacturing year (e.g. 2020): "))
kms = int(input("Enter kilometers driven: "))
owner = int(input("Enter number of owners: "))
fuel = input("Enter fuel type (Petrol/Diesel/CNG): ")
seller = input("Enter seller type (Individual/Dealer): ")
trans = input("Enter transmission (Manual/Automatic): ")
current_year = datetime.now().year
years_used = current_year - manufacture_year
fuel_map = {'Petrol': 0, 'Diesel': 1, 'CNG': 2}
seller_map = {'Individual': 0, 'Dealer': 1}
trans_map = {'Manual': 0, 'Automatic': 1}

fuel_enc = fuel_map[fuel]
seller_enc = seller_map[seller]
trans_enc = trans_map[trans]
user_input = np.array([[years_used, showroom_price, kms,owner, fuel_enc, seller_enc, trans_enc]])

warnings.filterwarnings("ignore")
predicted_price = rf.predict(user_input)

print(f"Predicted Selling Price: ₹ {predicted_price[0]:.2f} Lakhs")
```
